# 1. The Z-Test (One-Sample & Two-Sample)
#### **Scenario:**
#### A factory produces organic tea. The historical population mean weight is 250g with a known standard deviation of 12g. A batch of 40 packets is tested.

#### **A. Manual Formula Fitting (The "Hard" Way)**
We will calculate the Z-score and then find the P-value 

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import math
```

```python
# step-1: data & parameters
mu=250
sigma=12
n=40
sample_mean=246.5

# --- STEP 2: MANUAL FORMULA FITTING ---
# Formula: Z = (x_bar - mu) / (sigma / sqrt(n))

standard_error=sigma/np.sqrt(n)
z_score=(sample_mean-mu)/standard_error
z_score
```

```python
# Manual P-Value Calculation
# For a two-tailed test: 2 * (1 - Area under curve up to Z)
# P(Z>∣z∣)=1−P(Z≤∣z∣)
# p=2×P(Z>∣z∣)
z_standard_value=stats.norm.cdf(abs(z_score))
               
p_value_manual=2*(1-z_standard_value)
print('p_value',p_value_manual,'\n','z standard noramal value',z_standard_value)
```

```python
print(f"--- Manual Calculation ---")
print(f"Z-Statistic: {z_score:.4f}")
print(f"P-Value: {p_value_manual:.4f}") 

if p_value_manual < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

## Two-Sample Z-Test (Manual & Professional)
#### **Scenario:**
#### Comparing the average test scores of two different coaching centers (Center A and Center B). We know the population standard deviations are 10 and 12 respectively.

```python
# professional data
center_a=np.random.normal(75,10,120)
center_b=np.random.normal(78,12,150)
```

### 1. Manual Fitting
### Z = (X1bar - X2bar) / sqrt( (sigma1^2/n1) + (sigma2^2/n2) )


```python

mean_a,mean_b=np.mean(center_a),np.mean(center_b)
s1,s2=10,12
n1,n2=len(center_a),len(center_b)

z_2samp=(mean_a-mean_b)/np.sqrt((s1**2/n1)+(s2**2/n2))
z_2samp
```

```python
p_2samp=stats.norm.sf(abs(z_2samp))*2
p_2samp
```

```python
if p_2samp < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

###  2. Professional Approach

```python
z_stats_2,p_val_2=ztest(center_a,center_b)
```

```python
print(f"Two-Sample Z-Stat: {z_stats_2:.4f}, P-Value: {p_val_2:.4f}")

if p_val_2 < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

# 2. The T-Test (Independent & Paired)
#### **Scenario:** 
#### Testing if a "Memory Booster" supplement works. We compare "Group A" (Placebo) vs "Group B" (Supplement).



## A. Manual Formula Fitting (Independent T-test)

```python
group_a=[20,22,19,21,23]
group_b=[25,28,24,26,27]

# formulas: Mean,variance,Pooled SD and T-scores
mean1,mean2=np.mean(group_a),np.mean(group_b)
var1,var2=np.var(group_a,ddof=1),np.var(group_b,ddof=2)
n1,n2=len(group_a),len(group_b)

# Pooled standard deviation
pooled_se=math.sqrt((var1/n1)+(var2/n2))
t_manual=(mean1-mean2)/pooled_se
t_manual
```

```python
# Degree of freedom (simplified)
df=n1+n2-2
p_manual=stats.t.sf(abs(t_manual),df)*2

print(f"Manual T-Stat: {t_manual:.4f}, Manual P-Val: {p_manual:.4f}")

if p_manual < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

## One-Sample T-Test
#### **Scenario:**

#### A nutritionist claims a new diet reduces weight by 5kg in a month. We test this on 25 people. Since we don't know the population standard deviation, we use T-test.

```python
 # --- PROFESSIONAL DATA ---
weight_loss = np.random.normal(4.2, 1.5, 25) # Small sample, sigma unknown
weight_loss
```

```python
# 1. Manual Fitting
# t = (x_bar - mu) / (s / sqrt(n))
x_bar = np.mean(weight_loss)
mu_claim = 5
s = np.std(weight_loss, ddof=1) # Sample standard deviation
n = len(weight_loss)
```

```python
t_manual = (x_bar - mu_claim) / (s / np.sqrt(n))
df = n - 1
p_manual = stats.t.sf(abs(t_manual), df) * 2
p_manual
```

```python
print("T-Test Value:", t_manual)
print("p value:", p_manual)

if p_manual < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

```python
# 2. Professional Approach
t_stat, p_val = stats.ttest_1samp(weight_loss, 5)
```

```python
print(f"One-Sample T-Stat: {t_stat:.4f}, P-Value: {p_val:.4f}")


if p_val < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

# Chi-Square Test: The "Association" Test
#### **Business Scenario:** 
#### A Marketing Manager wants to know if the Subscription Plan (Basic, Gold, Platinum) chosen by a customer is related to their Region (North, South, East, West). If they are independent, the same marketing strategy works everywhere. If they are dependent, we need regional strategies.1. State the 

**1. HypothesesNull Hypothesis ($H_0$):** Subscription Plan and Region are Independent (No relationship).

**2. Alternative Hypothesis ($H_1$):** Subscription Plan and Region are Dependent (There is a significant relationship).

## A. Theoretical/Manual Calculation (Pen & Paper Logic)

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

observed = np.array([[150, 100, 50],   # North
                     [100, 150, 150]]) # South

# --- MANUAL STEP-BY-STEP ---

# A. Calculate Expected Values: (Row Total * Col Total) / Grand Total
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
grand_total = observed.sum()
expected = np.outer(row_totals, col_totals) / grand_total
row_totals,col_totals,grand_total,expected
```

```python
# B. Calculate Chi-Square Statistic: sum( (O - E)^2 / E )
chi_sq_manual = np.sum((observed - expected)**2 / expected)
chi_sq_manual
```

```python
# C. Calculate Degrees of Freedom (df): (Rows - 1) * (Cols - 1)
df_manual = (observed.shape[0] - 1) * (observed.shape[1] - 1)
df_manual
```

```python
# D. Manual P-Value: Survival Function of chi2 distribution
# This converts our "score" into a probability
p_value_manual = stats.chi2.sf(chi_sq_manual, df_manual)
p_value_manual # it is very-very less
```

```python
print("--- MANUAL RESULTS ---")
print(f"Chi-Square Stat: {chi_sq_manual:.4f}")
print(f"Degrees of Freedom: {df_manual}")
print(f"P-Value: {p_value_manual:.4e}") # Using scientific notation for small values
```

```python
if p_value_manual < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

## B. Professional Project Approach (Large Scale)
#### Imagine a dataset with 5,000 customers and 10 different features. We select the two categorical columns and test them.

```python
# Creating a Large Project Dataset
df_marketing = pd.read_csv(r"E:\data\data_set\df_marketing.csv")
df_marketing.head()
```

```python
# Step 1: Create Contingency Table
contingency_table = pd.crosstab(df_marketing['Region'], df_marketing['Plan'])
contingency_table
```

```python
# Step 2: The Scipy approach
chi2, p, dof, expected_table = stats.chi2_contingency(contingency_table)

print("\n--- PROFESSIONAL PROJECT RESULTS ---")
print(f"Chi2 Stat: {chi2:.2f}")
print(f"P-value: {p:.4f}")

# Step 3: Decision Logic
alpha = 0.05
if p < alpha:
    print("Decision: Reject Null Hypothesis (H0).")
    print("Insight: Region and Plan are DEPENDENT. Customize your marketing!")
else:
    print("Decision: Fail to Reject Null Hypothesis (H0).")
    print("Insight: Region and Plan are INDEPENDENT. Use a global strategy.")
```
