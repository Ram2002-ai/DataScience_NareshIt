# Assignment-1: Revenue Estimation

```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
df=pd.read_csv("E:\data\data_set\Revenue_forcasting_CI.csv")
```

```python
df
```

# population parameters (mean,std)

```python
pop_mean=df['Monthly_Revenue'].mean()
pop_mean
```

```python
# variance population
deviation=df['Monthly_Revenue']-pop_mean
```

```python
deviation
```

```python
deviationsquare=deviation**2
deviationsquare=sum(deviationsquare)
```

```python
variance=deviationsquare/df.shape[0]
```

```python
variance
```

# standard deviation of population

```python
std=np.sqrt(variance)
std
```

# sample of statestics(mean,std)

```python
np.random.seed(42)
sample=df.sample(60)
```

```python
sample
```

```python
sample_mean=sample['Monthly_Revenue'].mean()
sample_mean
```

# sample standard deviation

```python
sample_devaition=sample['Monthly_Revenue']-sample_mean
sample_devaition
```

```python
squra_sampl_deviation=sum(sample_devaition**2)
```

```python
sample_var=squra_sampl_deviation/sample.shape[0]
sample_var
```

```python
sample_std=np.sqrt(sample_var)
sample_std
```

```python
## varian by function
```

```python
df['Monthly_Revenue'].var()
```

```python
# variance by calculation
variance
```

```python
# std by function
df['Monthly_Revenue'].std()
```

```python
# std by calculation
std
```

```python
# all results 
print(f'pupulation mean is: {pop_mean}, and standard devaiton is: {std}')
print(f'sample mean is: {sample_mean}, and standard devaiton is: {sample_std}')
```

### Point Estimate of churn rate 
- point estimate means mean of the churn rate

```python
point_est=df['Churn'].mean()
point_est
```

## Q7: Construct 95% Confidence Interval for Proportion

# Sample Propertion
p=x/n , x - number of succefull outcomes ,n-total number of outcomes

```python
churn_1=df[df['Churn']==1]
churn_1
```

```python
sample_proportion=churn_1.shape[0]/df.shape[0]
sample_proportion
```

```python
standard_error=np.sqrt(sample_proportion*(1-sample_proportion)/df.shape[0])
standard_error
```

# Z-value
- standardization of values

```python
z_value=sum((df['Churn']-point_est)/df['Churn'].shape[0])
z_value
```

```python
z_score=0.9911
```

```python
margin_of_error=z_score*standard_error
margin_of_error
```

# Confidence Interval

```python
mini=point_est-margin_of_error
mini
```

```python
maxi=point_est+margin_of_error
maxi
```

```python
point_est
```

```python
# confidence interval
print(f'confidence interval is lower bound {mini} mean {point_est} upper bound{maxi}')
```
