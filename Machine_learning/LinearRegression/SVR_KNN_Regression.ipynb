# Support Vector Regression 
- Support Vector Regression is a supervised machine learning algorithm that predicts continuous values by fitting a function within a predefined error margin called epsilon while minimizing model complexity.

- formula

        -a<y=wx+b<+a
# Hyperplane

- The regression line that predicts the output.
 y=wx+b

# Support Vectors

- These are the data points closest to the boundary of the ε tube.

- They determine the position of the regression line.

- That is why they are called support vectors.

# Kernel Trick

- If data is not linear, SVR uses kernels to convert data to higher dimensions.

  - common kernels

     - Linear (linear data)
     - Polynomial (polynomial relationship)
     - RBF-Radial basis function (Non-linear complex data)
     - Sigmoid (neural network like)

# Important Parameters in SVR
- C (Regularization Parameter)

- Controls penalty for errors.

    Small C
    → larger margin
    → more errors allowed

    Large C
    → tries to fit data tightly
    → risk of overfitting


# ε (Epsilon)

- Defines acceptable error margin.

    Small ε
    → model tries to fit more points
    
    Large ε
    → smoother model

- Epsilon (ε)

- It defines the margin of tolerance.

Example:
    
    If ε = 1, prediction errors within ±1 are ignored.
    
    So SVR does not try to fit every point exactly.

# 7️⃣ Advantages

✔ Works well for high dimensional data

✔ Handles non-linear relationships

✔ Robust to outliers

# 8️⃣ Disadvantages

❌ Slow for large datasets

❌ Hard to choose best kernel and parameters

# KNN-Regressor
- KNN Regressor is a supervised machine learning algorithm used for regression problems (predicting continuous values like price, temperature, sales, etc.).

- Instead of learning a mathematical model, it predicts the output based on the nearest data points in the dataset.

- Simple Idea

- KNN Regressor predicts the value of a new data point by taking the average of the K nearest data points.

# 1️⃣ How KNN Regression Works

- Steps:

    1.Choose a value for K (number of neighbors).
    
    2. Calculate the distance between the new data point and all training data points.
    
    3. Select the K nearest points.
    
    4. Take the average of their target values.
    
    5. That average becomes the predicted value.

# 2️⃣ Distance Metrics

- KNN uses distance to find neighbors.

- Common distances:

    - Euclidean Distance (Most Used)

    d=sqrt((x1-x2)**2+(y1-y2)**2)

     - Manhattan Distance

     d=abs(x1-x2)+abs(y1-y2)

# 3️⃣ Important Parameters
- 1️⃣ n_neighbors (K)

- Number of neighbors used for prediction.
    - Small K → sensitive to noise
    - Large K → smoother predictions

- 2️⃣ weights

- How neighbors influence prediction.

   - uniform
    
    All neighbors have equal importance.
    
   - distance
    
    Closer neighbors have more influence.


# 5️⃣ When to Use KNN Regression

- Use when:

  -  Dataset is small
    
  -  Data pattern is non-linear
    
  -  No assumption about data distribution

# 6️⃣ Advantages

    ✔ Simple algorithm
    ✔ No training phase
    ✔ Works well with non-linear data

# 7️⃣ Disadvantages
    
    ❌ Slow for large datasets
    ❌ Sensitive to feature scaling
    ❌ Requires storing full dataset
    
# 8️⃣ Important Note (Very Important in Interviews)

- Feature Scaling is required for KNN because distance calculations are affected by -feature magnitude.

- Example scaling methods:

    Standardization
    
    MinMaxScaler

- KNN Regressor is a non-parametric supervised learning algorithm that predicts continuous values by averaging the target values of the K nearest data points based on a distance metric.

# knn-regressor code

```python
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor(n_neighbors=4,weights='distance',algorithm='ball_tree')
knn.fit(X,y)

knn_pred=knn.predict([[6.5]])
knn_pred
```

```python
from sklearn.metrics import r2_score
```

```python
knn_pred=knn.predict(X)
```

```python
score=r2_score(y,knn_pred)
score
```
