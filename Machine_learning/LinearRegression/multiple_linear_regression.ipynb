# Multiple Linear Regression
- Multiple Linear Regression (MLR) is a machine learning / statistical technique used to predict a continuous dependent variable using two or more independent variables.

- OR we can Say as:

    - Multiple Linear Regression models the relationship between one dependent variable (Y) and multiple independent variables (X₁, X₂, X₃ … Xₙ).

- Mathematical Equation :

      - Y= b0+b1x1+b2x2+b3x3+....+bnxn

  - Where:

        Y → Dependent variable (target/output)
        
        X₁, X₂, X₃ … → Independent variables (features)
        
        b₀ → Intercept (constant)
        
        b₁, b₂, b₃ … → Coefficients (weights of features)

#### Assumptions of Multiple Linear Regression

       - Linear Relationship between X and Y
        
       - No Multicollinearity (independent variables should not be highly correlated)
        
       - Homoscedasticity (constant variance of errors)
        
       - Normal distribution of residuals
        
       - Independence of observations

       
#### Evaluation Metrics

    - Common metrics used:
    
       - R² Score
        
       - Adjusted R²
        
       - Mean Absolute Error (MAE)
        
       - Mean Squared Error (MSE)
        
       - Root Mean Squared Error (RMSE

# Ordinary Least Square
-

- Ordinary Least Squares (OLS) is a method used in Linear Regression to find the best-fit line for predicting a dependent variable.

- It chooses the line that minimizes the sum of squared differences (errors) between the actual values and the predicted values.
- OLS finds β₀ and β₁ so that prediction error is minimum.

##### Error(Residual)
 - For each data points

    - Residual=Actual-Predicted
    - (ei=yi-yi^) 

##### Why "Squared"?

OLS minimizes:

- sse =(ei=yi-yi^)**2 sum of squared errors

- This is called Sum of Squared Errors (SSE).

- Reasons we square errors:

   - Negative errors become positive
    
   - Large errors are penalized more
    
   - Easier mathematical optimization

### Summary
- Ordinary Least Squares is a statistical technique used in linear regression to estimate model parameters by minimizing the sum of squared residuals between actual and predicted values.

```python
import statsmodels.api as sm
X_opt=X[:,[0,1,2,3,4,5]]
```

```python
""" Exogenous variables (independent variables / features).
- Endogenous variable (dependent variable).
- Train an Ordinary Least Squares regression model using y as the target variable and X_opt as the feature matrix."""

regressor_OLS=sm.OLS(endog=y,exog=X_opt).fit()
regressor_OLS.summary()
```

# Backward Elimination (Feature Selection)

- Backward Elimination is a feature selection technique used with Ordinary Least Squares (OLS) regression to remove statistically insignificant variables from a model.

- It uses p-values from statsmodels OLS results to decide which feature should be removed.

#### Algorithm Steps

- Start with all independent variables

- Fit OLS regression

- Check p-values

- Remove the variable with highest p-value

- Refit the model

- Repeat until all p-values < 0.05

```python
import statsmodels.api as sm
X_opt=X[:,[0,1,2,3,5]] #remove the 4 feature
# ols
regressor_ols=sm.OLS(endog=y,exog=X_opt).fit()
regressor_ols.summary()
```

# Bias  (High Bias)= Underfitting
- Bias is the difference between the model’s expected prediction and the true value due to incorrect assumptions in the learning algorithm.

- Bias is the error that occurs when a model is too simple to capture the true patterns in the data.

  -  High bias: The model oversimplifies, misses patterns and underfits the data.
  -  Low bias: The model captures patterns well and is closer to the true values.

- Bias^2= (predicted -actual)**2


# Variance (High Variance)=Overfitting
- Variance is the error caused by the model’s sensitivity to small fluctuations in the training data. High variance leads to overfitting.
- It fails to give prediction on unseen new data.
- High variance: The model is too sensitive to small changes and may overfit.
- Low variance: The model is more stable but might miss some patterns.
