# Regularization
- Regularization is a technique used in machine learning to prevent overfitting by adding a penalty term to the loss function so that the model does not become too complex.

 **It is commonly used in models like:**

    Linear Regression
    
    Logistic Regression
    
    Support Vector Machine
    
    Neural Networks using Weight Decay

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Preprocessing
- Data preprocessing is the process of cleaning, transforming, and preparing raw data into a structured format suitable for machine learning algorithms.

# Feature Scaling

- Feature Scaling is a data preprocessing technique used in Machine Learning to bring all features (variables) to a similar scale or range. 📊

- Many ML algorithms (like Linear Regression, KNN, SVM, Gradient Descent) work better when the input features have similar magnitudes.

# PolynomialFeatures
- PolynomialFeatures is a preprocessing technique used to convert linear features into polynomial features so that a linear model can learn non-linear relationships.

- It is commonly used in Polynomial Regression, which extends Linear Regression to model curved relationships.

# Ridge Regression
- Ridge Regression is a regularized version of Linear Regression that helps reduce overfitting by adding a penalty on large coefficients.

- Ridge Regression is a regularization technique that adds an L2 penalty to the loss function to shrink model coefficients and reduce overfitting.

- Advantages of Ridge Regression

✔ Reduces overfitting
✔ Handles multicollinearity
✔ Improves model stability
✔ Works well with many features

- Disadvantages

❌ Does not perform feature selection
❌ All features remain in the model

# Lasso Regression
- Lasso Regression is a regularization technique used to reduce overfitting and perform feature selection in regression models.

- Advantages of Lasso

✔ Reduces overfitting
✔ Performs automatic feature selection
✔ Produces simpler models
✔ Useful for high-dimensional data

- Disadvantages

❌ If features are highly correlated, Lasso may randomly select one and remove others
❌ May remove useful features in some cases

# R2_score
