```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv')
```

```python
df.head()
```

```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import cross_val_score,KFold
from sklearn.neighbors import KNeighborsRegressor
```

```python
knn=KNeighborsRegressor()
```

```python
kfold=KFold(n_splits=5,shuffle=True,random_state=1)
scores=cross_val_score(knn,X,y,cv=kfold,scoring='r2')
```

```python
scores.mean()
```

# GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
knn=KNeighborsRegressor()
```

```python
param_grid = {
    'n_neighbors':[1,3,5,7,10,12,15,17,20],
    'weights':['uniform','distance'],
    'algorithm':['ball_tree', 'kd_tree', 'brute'],
    'p':[1,2]
}
```

```python
gsv=GridSearchCV(knn,param_grid,scoring='r2',refit=True,cv=kfold,verbose=2)
```

```python
gsv.fit(X,y)
```

```python
gsv.best_params_
```

```python
gsv.best_score_
```

```python
gsv.cv_results_
```

```python
pd.DataFrame(gsv.cv_results_)[['param_algorithm','param_n_neighbors','param_p','param_weights','mean_test_score']].sort_values('mean_test_score',ascending=False)
```

```python
new_data=np.array([0.0634,19,	2.31,	0,	0.538,	6.575,	65.2,	4.0900,	1,	296,	15.3	,396.90,	4.98	]).reshape(1,-1)
```

```python
gsv.predict(new_data)
```
