# One-Hot-Encoding
### 1️⃣ What is One-Hot Encoding?

- One-Hot Encoding is a technique used to convert categorical variables into multiple binary (0/1) columns.

- Each category becomes a separate column, and the value is:

- 1 → category present

- 0 → category absent

- convert each category into one column if we have 50 category then it make 50 columns

- Ohe used on **Nominal Categorical column**

### 3️⃣ Why One-Hot Encoding is Used

- Machine learning models cannot understand text categories.


### 4️⃣ When to Use One-Hot Encoding

- Use it when:

  -  Categories have no order
    
  -  Feature is nominal


### 6️⃣ Dummy Variable Trap ⚠️

- When using One-Hot Encoding, sometimes one column should be dropped to avoid multicollinearity.


### 7️⃣ Advantages
    
    ✅ Prevents false ordering
    
    ✅ Works well for linear models
    
    ✅ Easy to implement

### 8️⃣ Disadvantages

    ❌ Creates many columns (curse of dimensionality)
    ❌ Not efficient for high-cardinality features


### 🔟 One-Line Interview Answer

- One-Hot Encoding converts categorical variables into binary columns so machine learning models can process nominal categorical data without introducing false ordering.
