# 1️⃣ What is Ordinal Encoding?

- Ordinal Encoding is a technique used to convert categorical data with an inherent order (ranking) into numerical values.

- Each category is assigned an integer value based on its order.

- we can identify which category is superier


## Ordinal Encoding only used for indipendent ordinal Categorical data


### 2️⃣ Why Ordinal Encoding is Needed

- Machine learning models cannot work with text values directly, so categorical variables must be converted into numbers.

- Ordinal encoding preserves the order relationship between categories.


### 3️⃣ When to Use Ordinal Encoding

- Use it when categories have a meaningful order.


### 4️⃣ When NOT to Use Ordinal Encoding

- Do not use it for nominal categories (no order).


### 6️⃣ Advantages

  -  ✅ Simple and fast
  -  ✅ Maintains category order
  -  ✅ Works well for tree-based models
        
###7️⃣ Disadvantages

    ❌ Assumes numerical relationship between categories
   
    ❌ Not suitable for nominal features


#### 9️⃣ One-Line Interview Answer

Ordinal Encoding converts ordered categorical variables into numerical values while preserving their ranking relationship.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day26-ordinal-encoding/customer.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.shape
```

```python
df=df.iloc[:,2:]
```

```python
df.head(3)
```

# Ordinal Encoder

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('purchased',axis=1), df['purchased'], test_size=0.33, random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
# we have to assign order manualy other wise the model choose random order
```

```python
oe=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
oe.fit(X_train)
```

```python
X_train_order=oe.transform(X_train)
X_test_order=oe.transform(X_test)
X_train_order
```

```python
oe.categories_
```

# Label Encoding
- Label Encoding converts categorical values into numerical labels so machine learning models can process categorical data.

#### label Encoder only used with dependent (target variable) data if data is ordinal catogorical

### When to Use Label Encoding

- Label encoding is best used when:
    
    - The feature is ordinal (has order)
    
    - The column is the target variable (y)


### When NOT to Use Label Encoding

- Avoid it for nominal categories (no order).

### Advantages
    
    ✅ Simple and fast

    ✅ Reduces memory usage

    ✅ Works well for tree-based models

### Disadvantages

    ❌ Introduces false order

    ❌ Can mislead algorithms like KNN, SVM, Linear Regression

* Label Encoder

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
le.fit(y_train)
```

```python
y_train_label=le.transform(y_train)
y_test_label=le.transform(y_test)
```
