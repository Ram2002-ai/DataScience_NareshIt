# Feature Scalling
- Feature scaling is a preprocessing technique used to transform numerical features into a similar range so that machine learning algorithms perform efficiently and no feature dominates due to large magnitude values.

### Why Scaling is Important

- Many ML algorithms are distance-based or gradient-based, so they require scaled data.

- Without scaling:

    - Model training becomes slow
    
    - Some features dominate the model
    
    - Accuracy may decrease
 
### Types of Feature Scaling
   - 1. Standardization (Z-score Scaling)
    
    Formula:
         Z= (X-muuu)/std

 - where:
   muu=mean
   sigma=standard deviation
 - Result
      mean=0 (mean always 0 of standardize values)
      standard devaition=1 (std always 1 of standardize values)
 - Used in:
    -  SVM/SVR
    -  Logistic Regression
    -  PCA

    - 2.  Normalization (Min-Max Scaling)

    - Formula:

      X'=(X-Xmin)/(Xmax-Xmin)
   
- Result:

- Values between 0 and 1

- Used in:

    Neural Networks
    
    Image processing
    
    Deep Learning

### 3. Robust Scaling

- Uses median and IQR, so it works well with outliers.


### 4️⃣ Algorithms That Require Scaling

- Need scaling:

    KNN
    
    K-Means
    
    SVM / SVR
    
    Logistic Regression
    
    Gradient Descent algorithms
    
    PCA

- Do NOT require scaling:

    Decision Tree
    
    Random Forest
    
    XGBoost

# Standardization
- Standardization is a feature scaling technique used in Machine Learning preprocessing to transform data so that it has:

   ### Formula:
  ####   Z= (X-muuu)/std

- Mean (μ) = 0

- Standard Deviation (σ) = 1

- It is also called Z-score normalization.

### Why Standardization is Important

- Many ML algorithms depend on distance or gradients, so features must be on the same scale.

    - example
     - suppose we have the dataset of age and salary 
     - the algorithm may give more importance to salary of larger values
     - standardization solve this

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("E:\data\data_set\Social_Network_Ads.csv")
```

```python
df.head()
```

```python
df=df.iloc[:,2:]
df.head()
```

# Train test split

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
X_train.shape,X_test.shape
```

# StandardScaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
# FIT the scaler to the train set ,it will learn the parameter
scaler.fit(X_train)
```

```python
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
X_train_scaled
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled)
X_test_scaled = pd.DataFrame(X_test_scaled)
X_test_scaled
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

ax1.scatter(X_train[0], X_train[1])
ax1.set_title("Before Scaling")
ax2.scatter(X_train_scaled[0], X_train_scaled[1],color='red')
ax2.set_title("After Scaling")
plt.show()
```
