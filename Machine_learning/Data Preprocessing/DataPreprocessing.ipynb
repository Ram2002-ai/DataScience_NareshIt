# Imputation

### Impute
- Transformers for missing value imputation.
- Impute means 👉 fill missing values (NaN) in your dataset.

### SimpleImputer
- Univariate imputer for completing missing values with simple strategies.

- Replace missing values using a descriptive statistic (e.g. mean, median, or most frequent) along each column, or using a constant value.

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer()
```

## Fit()
- "It learns the statistical values (mean/median/mode) from the training data, which will later be used to replace missing values."
- it only calculate the statestical values . not replace

```python
X
```

```python
imputer=imputer.fit(X[:,1:3])
```

```python
X
```

```python
imputer
```

## Transform()
- "It replaces missing values using the statistics learned during the fit stage."
-  fill the missing value in transform 
- with bydefault mean strategy
- The imputer has already learned the mean/median/mode.

```python
X[:,1:3]=imputer.transform(X[:,1:3])
```

```python
X
```

## Mediaan Strategy

```python
imputer1=SimpleImputer( missing_values='nan',strategy="median")
imputer1=imputer.fit(X[:,1:3])
imputer1
```

```python
X[:,1:3]=imputer.transform(X[:,1:3])
X
```

# Mode Strategy
- If “most_frequent”, then replace missing using the most frequent value along each column.
- Can be used with strings or numeric data. If there is more than one such value, only the smallest is returned.

```python
imputer2=SimpleImputer( missing_values='nan',strategy='most_frequent')
```

```python
imputer2
```

```python
imputer2=imputer.fit(X[:,1:3])
```

```python
X[:,1:3]=imputer.transform(X[:,1:3])
X
```

### Impute Categorical value for independent

## labelencoder
- LabelEncoder is a preprocessing technique used in Machine Learning to convert categorical (text) data into numerical labels.

- Machine learning models cannot understand text directly, so we convert categories into numbers.


🔹 Methods Used
    
        - fit() → Learn unique categories
        
        - transform() → Convert categories into numbers
        
        - fit_transform() → Do both in one step
        
        - inverse_transform() → Convert numbers back to original text

```python
from sklearn.preprocessing import LabelEncoder
```

```python
labelencoder_y=LabelEncoder()
# y=labelencoder_y.fit_transform(y)
labelencoder_y=labelencoder_y.fit(y)
labelencoder_y
y=labelencoder_y.transform(y)
y
```

## train_test_split

- train_test_split is a function from scikit-learn used to split your dataset into:

- Training data → Used to train the model

- Testing data → Used to evaluate the model

  #### Parameter Explained

| Parameter         | Meaning                                                    |
| ----------------- | ---------------------------------------------------------- |
| `X`               | Input features                                             |
| `y`               | Target variable                                            |
| `test_size=0.2`   | 20% data for testing                                       |
| `random_state=42` | Fixes randomness (so result is same every run)             |
| `shuffle=True`    | Shuffle data before splitting (default)                    |
| `stratify=y`      | Keeps class proportion same (important for classification) |
