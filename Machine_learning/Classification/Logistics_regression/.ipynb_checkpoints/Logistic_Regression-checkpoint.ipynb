# LOGISTIC REGRESSION
#### Logistic Regression is a classification algorithm used in Machine Learning to predict categorical outcomes (mostly binary: 0 or 1, Yes/No, True/False).  



### 🔥 2. Core Idea (VERY IMPORTANT)
- Step 1: Linear Combination

- We first compute a linear equation:

          z = b0 + b1x1 + b2x2 + ... + bnxn

- Step 2: Apply Sigmoid Function

- 👉 Converts output into probability

- 🔥 3. Sigmoid Function (Heart of Logistic Regression)
- Properties:

- Output range: (0,1)

- S-shaped curve

- Smooth and differentiable

- 👉 Behavior:

        z → +∞ ⇒ p → 1
        
        z → −∞ ⇒ p → 0

- 🔥 4. Decision Boundary

- 👉 A threshold (usually 0.5) used to classify:

- p ≥ 0.5 → Class 1

- p < 0.5 → Class 0

- 👉 Changing threshold affects:

        Precision
        
        Recall


### Types of Logistic Regression
    1. Binary Logistic Regression
    
          2 classes (0/1)
    
    2. Multinomial Logistic Regression
    
        Multiple classes (no order)
    
    3. Ordinal Logistic Regression
    
        Ordered categories (Low, Medium, High)

# OVR APPROACH
- One-vs-Rest (OvR) (also called One-vs-All) is a strategy used to convert a binary classification algorithm into a multiclass classifier.

### 🔹 1. Core Idea

- Instead of solving multiclass directly, OvR breaks it into multiple binary classification problems.

- 👉 For K classes, OvR creates K separate models.

- Each model learns:

    - “Is this class OR NOT?”

### 🔹 2. How It Works (Step-by-Step)

- Assume we have 3 classes:
    - 👉 A, B, C

- Step 1: Create Binary Labels
       Original Class	Model A	Model B	Model C
        A	1	0	0
        B	0	1	0
        C	0	0	1
- Step 2: Train Models

        Model 1 → A vs (B, C)
        
        Model 2 → B vs (A, C)
        
        Model 3 → C vs (A, B)

- Each model is trained independently.


### Step 3: Prediction

- For a new input:

- Each model gives a score/probability.

- Example:
| Model     | Output |
| --------- | ------ |
| A vs Rest | 0.7    |
| B vs Rest | 0.2    |
| C vs Rest | 0.1    |






### 5. Advantages of OvR ✅

    ✔ Simple and easy to implement
    ✔ Works with any binary classifier (SVM, Logistic Regression)
    ✔ Scales well for large datasets
    ✔ Faster than One-vs-One when classes are many


### Disadvantages ❌

    ❌ Class imbalance problem
    
    One class vs all others → data becomes skewed
    
    ❌ Overlapping regions
    
    Multiple classifiers may predict high scores
    
    ❌ Not always optimal
    
    Doesn’t consider relationships between classes

# SoftMax Logistic Regressin

- Softmax Logistic Regression is an extension of binary logistic regression used when we have more than 2 classes.


### 1. Core Idea

- Instead of training multiple models like OvR, Softmax:
        👉 Trains ONE single model
        👉 Outputs probabilities for all classes simultaneously
