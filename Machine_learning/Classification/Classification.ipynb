# Confusion Matrix
- A confusion matrix is a table used to evaluate the performance of a classification model.

- It shows how many prediction are correct and how many are incorrect

- 

- 




### 2️⃣ Four Components
    1. True Positive (TP)
        
        Model predicted Positive, and the actual value is also Positive.
        
        Example:
        Spam email → predicted Spam

    2. True Negative (TN)

        Model predicted Negative, and the actual value is also Negative.
        
        Example:
        Normal email → predicted Not Spam

    3. False Positive (FP)
        
        Model predicted Positive, but actual value is Negative.
        
        Example:
        Normal email → predicted Spam
        
   -  This is also called Type-1 Error.

    4. False Negative (FN)

        Model predicted Negative, but actual value is Positive.
        
        Example:
        Spam email → predicted Not Spam
        
   - This is also called Type-2 Error.

# Accuracy

- Accuracy is the most basic metric used to evaluate a classification model.

- It measures how many predictions the model got correct out of all predictions.

#### In other words
- Accuracy is a classification metric that measures the proportion of correct predictions made by a model. It is calculated as (True Positives + True Negatives) divided by the total number of predictions.

- 




### When Accuracy is Misleading (Important Interview Point)

- Accuracy does not work well with imbalanced datasets.

- Example:

- Dataset of 1000 emails

- |                 | Predicted Spam | Predicted Not Spam |
| --------------- | -------------- | ------------------ |
| Actual Spam     | 0              | 50                 |
| Actual Not Spam | 0              | 950                |



- If model predicts all emails as Not Spam

- Accuracy = 95%

- But the model cannot detect spam at all.

- This is why we also use:

  1.  Precision
    
  2.  Recall
    
  3.  F1 Score

# Precision
#### Precision measures how many predicted positive cases are actually positive.

- in other words
        Out of all the cases the model predicted as positive, how many were correct?


- Precision Formula

    Precision =TP / (TP + FP)
    
    Where:
    
    TP (True Positive) → Correctly predicted positive
    
    FP (False Positive) → Incorrectly predicted positive


  



### 🎯 When Precision is Important

- Precision is important when False Positives are costly.

- That means we want to ensure that when the model predicts Positive, it should be correct most of the time.

- Precision focuses on reducing False Positives (FP).



### Real-World Cases Where Precision Matters
- 1️⃣ Email Spam Detection 📧

    If precision is low:
    
        Important emails may be marked as Spam
        
        Example:
            
            Job offer email marked as spam ❌
            
            So we want high precision so that emails marked as spam are actually spam.



- 2️⃣ Fraud Detection 💳
    
   - If precision is low:
    
             Genuine transactions may be blocked as fraud
    
            Example:
            
               Your bank blocks your normal payment ❌
            
                High precision ensures that transactions marked as fraud are truly fraud.


- Precision is important in problems where False Positives are costly, such as spam detection, fraud detection, and medical diagnosis, because it ensures that predicted positive cases are actually correct.


### Interview Definition

- Precision is the ratio of correctly predicted positive observations to the total predicted positive observations. It measures the accuracy of positive predictions.

# Recal (Sensitivity)
#### Recall measures how many actual positive cases the model correctly identified.

- In simple words:

- Out of all the real positive cases, how many did the model successfully detect?

Recall focuses on reducing False Negatives (FN)





- 📌 Example (Interview Friendly)

  -  Suppose a Disease Detection Model.
    
  -  Actual patients with disease = 100
    
  -  Model predictions:
    
      -  TP = 80 → Correctly detected disease
        
      -  FN = 20 → Missed disease cases
        
      -  Recall = 80 / (80 + 20)
        = 0.80 (80%)
        
        ✅ Meaning:
        
        The model detected 80% of actual disease cases.
 

# 📌 When Recall is Important

- Recall is important when missing a positive case is dangerous.

    #### 1️⃣ Medical Diagnosis 🏥
    
   - If recall is low:
    
   - Sick patients may be predicted healthy
    
       - Example:
            
          -  Cancer detection
            
          -  Missing a patient = very dangerous
            
          -  So we want high recall.
        


# 20-Second Interview Answer

    Recall is the ratio of correctly predicted positive observations to all actual positives.
    
    It measures how well the model captures positive cases and is calculated as TP divided by (TP + FN).
    
    Recall is important in applications like disease detection, fraud detection, and security systems where missing a positive case can be critical.

# F1-Score
- F1-Score is the balance between Precision and Recall.

- It is the harmonic mean of Precision and Recall, which means it gives a single score that considers both false positives and false negatives.





# F1-Score Formula

- F1 Score =2 × (Precision × Recall) / (Precision + Recall)

    Where:
    
    - Precision = TP / (TP + FP)
    
    - Recall = TP / (TP + FN)
 
### 📌 Why F1-Score is Used

- F1-Score is useful when the dataset is imbalanced.

   - Example:
    
            Spam detection
            
            990 Normal emails
            
            10 Spam emails
            
            Accuracy may look high but the model may miss spam emails.
    
   - F1-Score balances Precision and Recall, so it gives a better evaluation.
 

### 📌 F1 Score Range

| F1 Score | Meaning       |
| -------- | ------------- |
| 1        | Perfect model |
| 0        | Worst model   |

- Higher F1-Score = Better model performance.


### 📌 When to Use F1-Score

    Use F1-Score when:
    
    Dataset is imbalanced
    
    Both False Positives and False Negatives are important
    
    Examples:
    
        Fraud detection
        
        Spam detection
    
        Medical diagnosis

### Interview One-Line Answer

- F1-Score is the harmonic mean of Precision and Recall and is used to balance both metrics, especially when the dataset is imbalanced.

# Specificity
- Specificity measures how well a model correctly identifies negative cases.

- In simple words:

    - Out of all actual negative cases, how many did the model correctly predict as negative?



### 📌 Example

- Suppose a disease detection model:

- Actual healthy patients = 200

- Model predictions:

    TN = 180 → Correctly predicted healthy
    
    FP = 20 → Healthy predicted as sick
    
    Specificity =180 / (180 + 20)
    
    = 180 / 200
    
    = 0.90 (90%)
    
    ✅ Meaning:
    
   - The model correctly identified 90% of healthy patients.

### 📌 Why Specificity is Important

- Specificity is important when False Positives must be minimized.

   - Examples
    
            1️⃣ Medical Testing 🏥
            
            Avoid diagnosing healthy people as sick.
            
            2️⃣ Fraud Detection 💳
            
            Avoid flagging normal transactions as fraud.
            
            3️⃣ Spam Detection 📧
            
            Avoid marking normal emails as spam.

### 20-Second Interview Answer

- Specificity is the proportion of actual negative cases that are correctly identified by the model. It is calculated as TN divided by (TN + FP) and measures how well the model avoids false positives.

# ROC (Receiver Operating Characteristic)
#### ROC (Receiver Operating Characteristic) Curve is a graphical plot that illustrates a model’s performance across all classification thresholds.

- It shows the trade-off between:
    
    True Positive Rate (TPR / Recall / Sensitivity)
    
    False Positive Rate (FPR)

  
  

# Key Metrics in ROC
#### a) True Positive Rate (TPR) / Recall / Sensitivity

    TPR = TP / (TP + FN)
    
    Measures how many actual positives are correctly detected.
    
    Higher TPR → Better detection of positives.

    X-Axis 

#### b) False Positive Rate (FPR)

    FPR = FP / (FP + TN)
    
    Measures how many actual negatives are incorrectly classified as positive.
    
    Lower FPR → Fewer false alarms.

    Y-Axis


### Why ROC is Important

- When a model outputs probabilities (e.g., 0.0–1.0), you choose a threshold to classify:

   - Example: Threshold = 0.5
    
        Probability ≥ 0.5 → Positive
        
        Probability < 0.5 → Negative

   - ROC plots TPR vs FPR for all possible thresholds.
 

### Example – Fraud Detection

- Suppose a bank wants to detect fraudulent transactions:

| Transaction Type         | Count | Model Prediction                          |
| ------------------------ | ----- | ----------------------------------------- |
| Fraud (Actual Positive)  | 20    | 18 detected (TP), 2 missed (FN)           |
| Normal (Actual Negative) | 80    | 10 flagged as fraud (FP), 70 correct (TN) |


- TPR = 18 / (18 + 2) = 0.9

- FPR = 10 / (10 + 70) = 0.125

- If we change the threshold:

   - Lower threshold → Detect more frauds (higher TPR) but more false alarms (higher FPR)
    
   - Higher threshold → Fewer false alarms (lower FPR) but may miss frauds (lower TPR)

- ROC shows all threshold choices in one curve.

# AUC

- AUC = Area Under the ROC Curve

- It measures the overall ability of a classification model to distinguish between positive and negative classes.

- Think of it as a single-number summary of how good your model is across all thresholds.



### Real-World Example – Fraud Detection

- Imagine a bank detecting fraudulent transactions:

    - Model outputs probabilities:

       - Transaction 1 → 0.9 fraud probability
        
       - Transaction 2 → 0.2 fraud probability

- ROC curve shows TPR vs FPR for all thresholds.

- AUC = 0.85 → Good model → most frauds detected while keeping false alarms low.


####Interview Tip

- “ROC curve visualizes a classifier’s ability to separate positive and negative classes across thresholds. The closer the curve is to the top-left corner, the better the classifier. AUC summarizes performance: 1 = perfect, 0.5 = random.”
