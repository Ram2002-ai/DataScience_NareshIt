# SVM
### 🔷 1. Introduction to SVM

- Support Vector Machine (SVM) is a powerful supervised machine learning algorithm used for:

    ✅ Classification (binary & multiclass)
    
    ✅ Regression (SVR)
    
    👉 It became popular in the 1990s because:
    
        Works well even with small datasets
        
        Handles high-dimensional data
        
        Provides clear decision boundaries

### 🔷 2. Core Idea (Intuition)

    👉 SVM tries to:
    
        Find the best boundary (hyperplane) that separates classes with maximum margin
    
    Example (2D case):
        
        Data points of two classes
        
        Many possible lines can separate them
        
        SVM chooses the best line with maximum margin
