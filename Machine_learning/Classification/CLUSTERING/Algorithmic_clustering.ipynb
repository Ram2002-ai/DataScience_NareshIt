```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/campusx-official/agglomerative-hierarchical-clustering-demo/main/hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv"

df = pd.read_csv(url, encoding='latin1')
df.head()
```

```python
df.shape
```

```python
data=df.iloc[:,3:5].values
```

```python
data
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize=(10,7))
plt.title('Customer Dendograms')
dend=shc.dendrogram(shc.linkage(data,method='ward'))
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize=(10,7))
plt.title('Customer Dendograms')
dend=shc.dendrogram(shc.linkage(data,method='average'))
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize=(10,7))
plt.title('Customer Dendograms')
dend=shc.dendrogram(shc.linkage(data,method='median'))
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=5, linkage='ward')
labels_=cluster.fit_predict(data)
```

```python
labels_
```

```python
plt.figure(figsize=(10,7))
plt.scatter(data[:,0],data[:,1],c=cluster.labels_,cmap='rainbow')
```
