# 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

# Scatter Plot

# axes level function

```python
# axes level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

# Relplot -> Figure level 
- it has access of full figure
- square shape

```python
# hence it has full control of figure thats why it able to put legend outside the graph
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
plt.show()
```

# 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

- figure level -> displot
- axes level -> histplot -> kdeplot->rugplot

# plotting univariate histogram

```python
# axes level
sns.histplot(data=tips,x='total_bill')
```

```python
# figure level
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

- It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
- Discrete bins are automatically set for categorical variables

- countplot


```python
sns.displot(data=tips, x='day', kind='hist') # count of customer by day
```

```python
# hue parameter
sns.displot(data=tips, x='day', kind='hist',hue='sex') 
```

```python
# element->step
sns.displot(data=tips, x='day', kind='hist',hue='sex',element='step')# element -> bars,poly
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',element='poly',hue='sex')
```

# faceting using col and row
- not work on histplot function because of axes level function

```python
sns.displot(data=tips,x='tip',kind='hist',col='sex',element='step')
```

# KDE Plot
- Rather than using discrete bins ,a KDE plot smooths

```python
# axes level
sns.kdeplot(data=tips,x='total_bill')
```

```python
# figure level
sns.displot(data=tips,x='total_bill',kind='kde')
```

# Rugplot

```python
sns.rugplot(data=tips,x='total_bill')
```

# Heatmap
- Plot rectungle data as a color encoded matrix
- find the correlation b/w two variable

```python
temp=gap.pivot(index='country',columns='year',values='lifeExp')

# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp)
```

```python
# annot
temp1=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp1,annot=True,linewidth=0.5,cmap='summer')
```

# Boxplot
- A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”).
-  It can tell you about your outliers and what their values are.
- Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

```python
# axes level
sns.boxplot(data=tips,x='day', y='total_bill',hue='sex')
```

# Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
# axes level
# hue parameter is not avaible 
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
# figure level
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

# residplot
- tells the how far data points from the mean

```python
sns.residplot(data=tips,x='total_bill',y='tip')
```

# A second way to plot Facet plots -> FacetGrid

```python
# figure level->regplot,displot->catplot->lmplot
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
g = sns.FacetGrid(data=tips,col='day',row='time',hue='smoker')
g.map(sns.boxplot,'sex','total_bill')
g.add_legend()
```

```python
g = sns.FacetGrid(data=tips,col='day',row='time',hue='smoker',height=20,aspect=2)
g.map(sns.boxplot,'sex','total_bill')
g.add_legend()# we have to adjust height and aspect
```

# Ploting Pairwise Relationship (PairGrid vs Pairplot)

```python
iris=sns.load_dataset('iris')
```

```python
sns.pairplot(iris,hue='species')
```

```python
# pair grid
g=sns.PairGrid(data=iris,hue='species')
# g.map
g.map(sns.scatterplot)
```

```python
# map_diag->map_offdiag
g=sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
```

```python
# map_diag->map_upper->map_lower
g=sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g=sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

# JointGrid vs Jointplot

```python
# axes level
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist',hue='sex')
```

```python
g=sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.violinplot)
```

```python
g=sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.boxplot)
```
