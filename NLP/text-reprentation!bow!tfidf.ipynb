### removing the html tags, special chracters and stopwords

```python
import pandas as pd
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def clean_text(text):
    # 1. remove html tags
    text=re.sub(r'<.*?>','',text) 
    
    # 2.lowercase
    text=text.lower()

    # 3. remove special characters & numbers
    text=re.sub(r'[^a-zA-Z\s]',' ',text)

    # 4. Tokenization
    tokens=word_tokenize(text)

    # 5. Remove stopwords
    tokens=[word for word in tokens if word not in stop_words]

    # 6. Lemmatization
    tokens=[lemmatizer.lemmatize(word,pos='v') for word in tokens]

    

    return " ".join(tokens)
```

```python
df['clean_review']=df['review'].apply(clean_text)
```

# text preprocessing
### Grams
- unigram
- bigram
- trigram
- ngram
  

```python
from nltk.util import bigrams
def generate_bigrams(text):
    return [' '.join(bg) for bg in  bigrams(text)]
df['bigram']=df['word_tokenize'].apply(generate_bigrams)
df.head()
```

```python
len(df['bigram'][0])
```

```python
from nltk.util import trigrams
def generate_trigrams(text):
    return [' '.join(bg) for bg in  trigrams(text)]
df['trigram']=df['word_tokenize'].apply(generate_trigrams)
df.head()
```

```python
len(df['trigram'][0])
```

```python
from nltk.util import ngrams
def generate_ngrams(text):
    return [' '.join(bg) for bg in  ngrams(text,10)]
df['ngram']=df['word_tokenize'].apply(generate_ngrams)
df.head()
```

# OHE(One hot Encoding)
- 👉 In NLP, OHE is used differently:

- Instead of categories → we encode words

### But in NLP we DON'T use OHE directly ❗

- 👉 Because:

- Huge vocabulary → too many columns
- Sparse matrix (inefficient)

### 🔹 Instead use:
- Bag of Words
- TF-IDF

- (These are optimized versions of OHE for text 💯)

```python
from sklearn.preprocessing import OneHotEncoder


ohe=OneHotEncoder(sparse_output=False)
encoded=ohe.fit_transform(df[['clean_review']])
encoded_df=pd.DataFrame(encoded,columns=ohe.get_feature_names_out())
df_final=pd.concat([df,encoded_df],axis=1)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Bigram
cv_bigram = CountVectorizer(ngram_range=(2,2), max_features=5000)

X_bigram = cv_bigram.fit_transform(df['clean_review'])

bigram_vocab = cv_bigram.get_feature_names_out()

print("Bigram Vocabulary Size:", len(bigram_vocab))
```

```python
# Trigram
cv_trigram = CountVectorizer(ngram_range=(3,3), max_features=5000)

X_trigram = cv_trigram.fit_transform(df['clean_review'])

trigram_vocab = cv_trigram.get_feature_names_out()

print("Trigram Vocabulary Size:", len(trigram_vocab))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X=tfidf.fit_transform(df['clean_review'])
```

```python
vocab=tfidf.get_feature_names_out()
len(vocab)
```

# get idf scores

```python
idf_scores=tfidf.idf_

# combine word +idf
idf_dict=dict(zip(vocab,idf_scores))

# print top 10
for word,score in list(idf_dict.items())[:10]:
    print(word,':',score)
```
