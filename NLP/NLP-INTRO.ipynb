# Types of Tokens
1. Bigram -token with 2 words
2. Trigram- token with 3 words
4. Ngram - token with more than 3 words

```python
from nltk.util import bigrams,trigrams,ngrams
```

```python
string='Hello! Datascience . I am the most challenging Man in the world?'
quotes_tokens=nltk.word_tokenize(string)
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams=list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
dict(nltk.bigrams(quotes_tokens))
```

```python
quote_trigram=list(nltk.trigrams(quotes_tokens))
quote_trigram
```

```python
quote_ngram=list(nltk.ngrams(quotes_tokens,5))
quote_ngram
```

# Stemming-stem
- stem gives root form of the word

```python
from nltk.stem import PorterStemmer
```

```python
pst=PorterStemmer()
```

```python
pst.stem('goodness')
```

```python
pst.stem('opening')
```

```python
words_stem=['give','giving','given','slow','slowness']
pst.stem(words_stem)
```

```python
for word in words_stem:
    print(word+ ': '+pst.stem(word))
```

```python
from nltk.stem import LancasterStemmer
lst=LancasterStemmer()

for word in words_stem:
    print(word + ": " + lst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sbst=SnowballStemmer('english')
for word in words_stem:
    print(word + ": " + sbst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sbst=SnowballStemmer('german')
for word in words_stem:
    print(word + ": " + sbst.stem(word))
```

```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
word_len=WordNetLemmatizer()
```

```python
words_stem
```

# POS-Part of Speech
- pos has tags and description
- recognize grammer info

```python
sent='my name is ram sevaiwar'
sent_tokens=word_tokenize(sent)
sent_tokens
```

```python
for token in sent_tokens:
    print(nltk.pos_tag([token]))
```

```python
sent2='John is eating delicious cake'
sent_tokens=word_tokenize(sent2)

for token in sent_tokens:print(nltk.pos_tag([token]))
    
```
