# Plotly Basics

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.offline as pyo

matches=pd.read_csv(r"E:\data\matplotlib_data\matches.csv")
deliveries=pd.read_csv("E:\data\matplotlib_data\deliveries.csv")
```

```python
ipl=deliveries.merge(matches,left_on='match_id',right_on='id')
ipl.head()
```

```python
ipl.columns
```

# Scatter Plot

```python
# top50 batters
```

```python
top50=ipl.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(50).index.tolist()
new_ipl=ipl[ipl['batter'].isin(top50)]
new_ipl.head()
```

```python
# calculating SR
# sr=(sum_score/sum_ball)*100
runs=new_ipl.groupby('batter')['batsman_runs'].sum()
balls=new_ipl.groupby('batter')['batsman_runs'].count()

sr=(runs/balls)*100
sr=sr.reset_index()
sr
```

```python
# calculating avg
# avg=sum_runs/sum_outs
out=ipl[ipl['player_dismissed'].isin(top50)]
nout=out['player_dismissed'].value_counts()

avg=runs/nout
avg=avg.reset_index()
avg.rename(columns={'index':'batter',0:'avg'},inplace=True)
avg=avg.merge(sr,on='batter')
avg
```

```python
# scatter plot
trace=go.Scatter(x=avg['avg'],y=avg['batsman_runs'],mode='markers',
                 text=avg['batter'],marker={'color':'#00a65a','size':16})
data=[trace]
layout=go.Layout(title='Batsman Abg vs Sr',
                xaxis={'title':'Batsman Average'},
                yaxis={'title':'Batsman strike rate'})
fig=go.Figure(data=data,layout=layout)
pyo.plot(fig,filename='ram.html')
```

```python
# year by year batsman performance
single=ipl[ipl['batter']=='V Kohli']
performance=single.groupby('season')['batsman_runs'].sum().reset_index()
performance
```

# Line Plot

```python
trace=go.Scatter(x=performance['season'],y=performance['batsman_runs'],
                mode='lines',marker={'color':'#00a66a'})
data=[trace]
layout=go.Layout(title='year by year performance',
                xaxis={'title':'season'},
                yaxis={'title':'total runs'})
fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```

```python
# multiple line plot
# year by year batsman performance
single1=ipl[ipl['batter']=='RG Sharma']
performance1=single1.groupby('season')['batsman_runs'].sum().reset_index()
performance1
```

```python
trace1=go.Scatter(x=performance1['season'],y=performance1['batsman_runs'],
                mode='lines',marker={'color':'#00a66a'},name='RG Sharma')
trace2=go.Scatter(x=performance['season'],y=performance['batsman_runs'],
                mode='lines',name='V Kohli')
data=[trace1,trace2]
layout=go.Layout(title='year by year performance',
                xaxis={'title':'season'},
                yaxis={'title':'total runs'})
fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```

```python
def batsman_comp(*name):
    data=[]
    for i in name:
        single=ipl[ipl['batter']==i]
        performance=single.groupby('season')['batsman_runs'].sum().reset_index()

        trace=go.Scatter(x=performance['season'],y=performance['batsman_runs'],
                        mode='lines + markers',name=i)
        data.append(trace)

        layout=go.Layout(title='Batsman Record comparision',
                        xaxis={'title':'season'},
                        yaxis={'title':'Runs'})
        fig=go.Figure(data=data,layout=layout)
        pyo.plot(fig)
```

```python
batsman_comp('V Kohli','RG Sharma','MS Dhoni','G Gambhir')
```

```python
# top 10 batsman
top10=ipl.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10).index.tolist()
top10_df=ipl[ipl['batter'].isin(top10)]


top10_scores=top10_df.groupby('batter')['batsman_runs'].sum().reset_index()
top10_scores
```

```python
trace=go.Bar(x=top10_scores['batter'],y=top10_scores['batsman_runs'])
data=[trace]
layout=go.Layout(title='Top Runs Scorer in Ipl',
                xaxis={'title':'batsman name'},
                yaxis={'title':'runs'})
fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```
