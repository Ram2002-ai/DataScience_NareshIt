```python
def generate_plot(plot_type):
    fig=plt.figure(figsize=(8,5))

    if plot_type=='bar plot':
        plt.bar(colors,children,color=colors)
        plt.title('Bar plot of childrens')
        plt.xlabel('colors')
        plt.ylabel('no of liking children')

    elif plot_type=='line plot':
        plt.plot(colors,children,color='r')
        plt.title('line chart of children likes colors')
        plt.xlabel('colors')
        plt.ylabel('childrens')

    elif plot_type=='scatter plot':
        plt.scatter(colors,children,color='b')
        plt.title('scatter plot of children likes the colors')
        plt.xlabel('colors')
        plt.ylabel('chidrens')
        plt.grid(True)

    elif plot_type=='pie plot':
        plt.pie(children,labels=colors,autopct='%1.1f%%')
        plt.title('colors liking by chidren in percentages')
    
    elif plot_type=='histogram':
        plt.hist(children,color='purple')
        plt.title('hist plot')
        plt.xlabel('colors')
        plt.ylabel('childrens')
    
    plt.tight_layout()
    return fig

# Gradio UI

demo=gr.Interface(
    fn=generate_plot,
    inputs=gr.Radio(['bar plot','line plot','scatter plot','histogram','pie plot'],label='choose plot type'),

    outputs=gr.Plot(label='Visualization'),
    title='Children vs color visual liking',
    description='choose a chart type to visualize the data.'
)

demo.launch()
```
