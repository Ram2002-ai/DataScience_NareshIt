### print prime numbers till 100

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n*0.5)+1):
        if n%i==0:
            return False
    return True
print(is_prime(17))
x=[x for x in range(2,101) if is_prime(x)]
```

```python
def second_largest(num):
    largest=float('-inf')
    second=float('-inf')
    if len(num)<2:
        return 'list does not have second number'
    for n in num:
        if n>largest:
            second=largest
            largest=n
        elif largest<n>second:
            second=n
    return second
print(second_largest([12,56,76,76]))
print(second_largest([12]))
```

### count frequency of word

```python
def word_frequency(sentence):
    sentence=sentence.split()
   
    frequency={}
    for word in sentence:
        frequency[word]=frequency.get(word,0)+1
    return frequency
print(word_frequency('sumit is going for global sumit in india'))
```

### count frequency of character in word

```python
def char_frequency(word):
    frequency={}
    for char in word:
        frequency[char]=frequency.get(char,0)+1
    return frequency
print(char_frequency('sonamisagirl'))
```

### find the duplicate 

```python
def is_duplicated(word):
    seen=set()
    duplicated=set()
    for char in word:
        if char in seen:
            duplicated.add(char)
        else:
            seen.add(char)
    return duplicated
print(is_duplicated('saurabh is a man'))
```

### two sum

```python
def find_sum(num,target):
    seen=set()
    for n in num:
        if (target-n) in seen:
            return n,target-n
        seen.add(n)
print(find_sum([12,67,8,20,0],20))
```

### find largest number between 3 numbers

```python
def largest(a,b,c):
    if b<=a>=c:
        return a
    elif a<=b>=c:
        return b
    else:
        return c
print(largest(1,1,1))
print(largest(2,2,1))
print(largest(1,3,1))
```

### check whether two string is anagram

```python
def is_anagram(a,b):
    a=sorted(a)
    b=sorted(b)
    if a==b:
        return True
print(is_anagram('mona','namo'))
```

```python
def is_anagram(a,b):
    if len(a) != len(b):
        return False

    frequency={}
    for char in a:
        frequency[char]=frequency.get(char,0)+1

    for char in b:
        if char not in frequency:
            return False
      

        if frequency[char]<0:
            return False
        return True
print(is_anagram('monna','nanmo'))
```

```python
# armstrong numbers
def armstrong(num):
    temp=num
    digits=len(str(num))
    sum=0
    while temp>0:
        n=temp%10
        sum+=n**digits
        temp=temp//10
    if num==sum:
        print('armstrong')
    else:
        print('not a armstrong')
print(armstrong(153))
```
