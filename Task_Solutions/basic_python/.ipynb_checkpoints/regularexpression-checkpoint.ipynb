# 2. Search Operations

# Match():
- re.match() checks only at the beginning of the string.
- It tries to match the pattern '456' at position 0.
- it returns None if the string pattern is not at the begining

```python
test_string='123abc45678abcABC'
print(re.match(r'abc',test_string))
```

```python
print(re.match(r'123',test_string))
```

```python
print(re.match(r'456',test_string))
```

# Span()
- span() returns a tuple showing the start and end index of the matched substring.

```python
pattern = re.compile(r"123")

match = re.match(pattern, test_string)
print(match)
```

```python
match.span()
```

```python
start_index,end_index=match.span()
test_string[start_index:end_index]
```

```python
# start()- returns the start index
match.start()
```

```python
# end()- returns the end (exclude) index of the matching substring
match.end()
```

# group()
- returns the actual matched text

```python
match.group()
```

# re.Search
- re.search() looks for a pattern anywhere in the string.

```python
test_string
```

# re.compile() 

- re.compile() is used to create a regex pattern object that you can reuse multiple times.

- 👉 It improves performance when using the same pattern again and again.

```python
pattern=re.compile(r'abc')
match=re.search(pattern,test_string)
match # here 6 is exclude
```

```python
test_string[3:6]
```

```python
match.group()
```

```python
pattern.search(test_string)
```

```python
start_index,end_index=match.span()
test_string[start_index:end_index]
```

# re.findall()
- re.findall() returns all matches of a pattern in a string.

- 👉 Output is a list
- 👉 If no match → returns empty list [] (not None)

```python
match=re.findall(pattern,test_string)
match
```

# re.finditer()

- re.finditer() returns an iterator of Match objects for all matches in a string.

- 👉 Unlike findall() (which returns a list of strings),
- finditer() returns match objects, so you can access:

- group()

- start()

- end()

- span()

```python
pattern=re.compile(r'abc')
matche=re.finditer(pattern,test_string)
for mat in matche:
    print(mat)
```

```python
matche=re.finditer(pattern,test_string)
for mat in matche:
    print("Matched Text:", mat.group())
    print("Start:", mat.start())
    print("End:", mat.end())
    print("Span:", mat.span())
    print("-----")
```

# 3. Meta Characters

```python
string="123abc.456789abcdefGHI.ABC \n xyz345WXYZ"
string1="The cat sat on the mat with a bat and a hat."
string3="The cat sat on the mat with a bat and a hat. $at, 8at, at."
```

```python
text="""Hello World!
This is a test.
hello again!
Python is fun.
Hello Python"""
```

# dot(.)
- In regular expressions, the dot (.) means:

    - ✅ Match every single character
    - ❌ Except newline (\n) by default

```python
pattern=re.compile(r'.')
matches=re.finditer(pattern,string)

for i in matches:
    print(i)
```

```python
print(string1,end='\n\n')

pattern=re.compile(r'.')
matches=re.finditer(pattern,string1)
for i in matches:
    print(i)
```

```python
print(string3, end="\n\n")

pattern = re.compile(r".")

matches = re.finditer(pattern, string3)

for match in matches:
	print(match)
```

# carat(^)

```python
print(text,end='\n\n')
pattern=re.compile(r'hello')
matches=re.carat(pattern,text)

for i in matches:
    print(i)
```
