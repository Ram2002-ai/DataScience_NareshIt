# GroupBy Object

```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv("E:\data\matplotlib_data\imdb-top-1000.csv")
```

```python
movies.head()
```

# Built in Aggregation Functions on Groupby Objct

```python
genre=movies.groupby('Genre')[['Runtime','IMDB_Rating','No_of_Votes','Gross','Metascore']]
genre.std()
```

# Top 3 Genres by earnings

```python
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

# find the genre with highest avg IMDB

```python
movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

# Find the director with most popularity

```python
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie of each genre
movies.groupby('Genre')['IMDB_Rating'].max()
```
