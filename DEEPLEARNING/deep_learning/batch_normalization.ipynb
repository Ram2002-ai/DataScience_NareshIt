```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
```

```python
X,y=make_circles(n_samples=1000,noise=0.03,random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
from tensorflow.keras.layers import BatchNormalization
```

```python
model=Sequential()
model.add(Dense(2,activation='relu',input_dim=2))
model.add(Dense(2,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history1=model.fit(X,y,epochs=200,validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y,clf=model,legend=2)
plt.show()
```

# Batch Normalization

```python
model=Sequential()
model.add(Dense(2,activation='relu',input_dim=2))
model.add(BatchNormalization())
model.add(Dense(2,activation='relu'))
model.add(BatchNormalization())
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history2=model.fit(X,y,epochs=200,validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y,clf=model,legend=2)
plt.show()
```

```python
plt.plot(history1.history['loss'],label='model1')
plt.plot(history2.history['loss'],label='model2')
plt.legend()
plt.show()
```

```python
plt.plot(history1.history['accuracy'],label='model1')
plt.plot(history2.history['accuracy'],label='model2')
plt.legend()
plt.show()
```

```python
plt.plot(history1.history['val_accuracy'],label='model1')
plt.plot(history2.history['val_accuracy'],label='model2')
plt.legend()
plt.show()
```
