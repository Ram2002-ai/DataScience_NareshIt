# Text Over The image

```python
import cv2

img_get = cv2.imread("E:/Photos/horse.jpg")

if img_get is None:
    print("Error: Image not loaded")
else:
    txt = cv2.putText(
        img=img_get,
        text='7 horse',
        org=(30, 80),
        fontFace=cv2.FONT_HERSHEY_DUPLEX,
        fontScale=3,
        color=(0, 0, 255),
        thickness=3,
        lineType=cv2.LINE_8,
        bottomLeftOrigin=False
    )

    txt = cv2.putText(
        img=img_get,
        text='7 horse',
        org=(80, 80),
        fontFace=cv2.FONT_HERSHEY_DUPLEX,
        fontScale=3,
        color=(144, 0, 114),
        thickness=3,
        lineType=cv2.LINE_8,
        bottomLeftOrigin=True
    )


    cv2.imshow('7 horse', txt)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

# Draw Line And Rectangle On Image

```python
import cv2
old_img=cv2.imread("E:\Photos\mohit.jpeg")
old_img=cv2.resize(old_img,(500,700))

new_img=cv2.line(img=old_img,pt1=(170,160),pt2=(350,150),color=(0,255,0),thickness=4,lineType=4)
cv2.imshow('Mohit',new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
old_img=cv2.imread("E:\Photos\mohit.jpeg")
old_img=cv2.resize(old_img,(500,700))
img_txt=cv2.putText(img=old_img,text='Mohit. 2 dino se lapata hai',org=(30,30),fontFace=2,fontScale=1,color=(255,55,2),thickness=2,lineType=16,bottomLeftOrigin=False)
new_img=cv2.rectangle(img=img_txt,pt1=(100,50),pt2=(380,500),color=(0,255,0),thickness=4,lineType=16)
cv2.imshow('Mohit',new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#  Draw a circle or Ellipse


```python
import cv2
old_img=cv2.imread("E:\Photos\mohit.jpeg")
old_img=cv2.resize(old_img,(500,600))

new_img=cv2.circle(img=old_img,center=(200,120),radius=150,color=(0,225,0),thickness=4,lineType=16)
cv2.imshow('mohit',new_img)
cv2.waitKey(0)
```

```python
import cv2
old_img=cv2.imread("E:\Photos\mohit.jpeg")
old_img=cv2.resize(old_img,(500,600))

new_img=cv2.ellipse(img=old_img,center=(200,120),radius=150,color=(0,225,0),thickness=4,lineType=16)
cv2.imshow('mohit',new_img)
cv2.waitKey(0)
```
