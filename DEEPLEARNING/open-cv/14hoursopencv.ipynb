# Image Scaling, Rotating using OpenCV  

```python
import cv2
import numpy as np
# Fix path using raw string
img1 = cv2.imread(r"E:\Photos\mohit.jpeg")

if img1 is None:
    print("Error: Image not found!")
else:
    img1 = cv2.resize(img1, (500, 500))
    print(img1.shape)

    h, w = img1.shape[:2]

    # FIX: center, angle, scale should be separate arguments
    matrix = cv2.getRotationMatrix2D((w/2, h/2), 180, 1)

    # Apply rotation
    rotated = cv2.warpAffine(img1, matrix, (w, h))
    h=np.hstack((img1,rotated))

    # Show rotated image (not original)
    cv2.imshow('Rotated Image',h)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

 # Image blurring using OpenCV

```python
import cv2
import numpy as np
# Fix path using raw string
img1 = cv2.imread(r"E:\Photos\mohit.jpeg")
img1 = cv2.resize(img1, (300, 300))

b=cv2.GaussianBlur(img1,(5,5),0)
m=cv2.medianBlur(img1,5)
c=cv2.bilateralFilter(img1,9,75,75)


h=np.hstack((img1,b,m,c))

cv2.imshow('Rotated Image',h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# MakeBorder using OpenCV

```python
import cv2
import numpy as np
# Fix path using raw string
img1 = cv2.imread(r"E:\Photos\mohit.jpeg")
img1 = cv2.resize(img1, (340, 340))

copy=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_CONSTANT,None,value=2)
copy=cv2.resize(copy,(340,340))

copy1=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_DEFAULT,None,value=2)
copy1=cv2.resize(copy,(340,340))

copy2=cv2.copyMakeBorder(img1,20,20,20,20,cv2.BORDER_REFLECT,None,value=2)
copy2=cv2.resize(copy,(340,340))

h=np.hstack((img1,copy,copy1,copy2))

cv2.imshow('mohit',h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Morphological operations using OpenCV

```python
import cv2
import numpy as np
```

- Erosion - shrink the channels
- dilation - sparse the channels
- opening - remove the noise
- closing
- gradient -differnce b/w errosion and dilation
- top hat
- black hat

```python
img=cv2.imread("E:\Photos\horse.jpg")
img=cv2.resize(img,(500,500))
m=np.ones((5,5),np.int8)
er=cv2.erode(img,m,iterations=1)
di=cv2.dilate(img,m,iterations=1)

op=cv2.morphologyEx(img,cv2.MORPH_OPEN,m,iterations=1)

cl=cv2.morphologyEx(img,cv2.MORPH_OPEN,m,iterations=1)

gr=cv2.morphologyEx(img,cv2.MORPH_GRADIENT,m,iterations=1)

tp=cv2.morphologyEx(img,cv2.MORPH_TOPHAT,m,iterations=1)

bp=cv2.morphologyEx(img,cv2.MORPH_BLACKHAT,m,iterations=1)

cv2.imshow('image',img)
# cv2.imshow('er',er)
# cv2.imshow('dil',di)
# cv2.imshow('opening',op)
# cv2.imshow('closing',cl)
# cv2.imshow('gradient',gr)
# cv2.imshow('tophat',tp)
cv2.imshow('tophat',bp)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Background subtraction using OpenCV

```python
import cv2
```

```python
org_v=cv2.VideoCapture(r"C:\Users\USER\Videos\Captures\pretrained_model_transformers.ipynb - Colab - Google Chrome 2026-04-24 00-01-00.mp4")
sub_m=cv2.createBackgroundSubtractorMOG2()
while True:
    r,frame=org_v.read()
    if r==True:
        frame=cv2.resize(frame,(500,700))
        sub_v=sub_m.apply(frame)
        cv2.imshow('video',frame)
        cv2.imshow('sub',sub_v)
        if cv2.waitKey(25) & 0xff==ord('p'):
            break
            
    else:
        break
org_v.release()
cv2.destroyAllWindows()
```
