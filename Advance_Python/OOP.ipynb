# ENCAPSULATIONS
### OBJECTS:
- type of objects:
          1. variable
          2. Methods
### Variable
    - instance variable (The var which we defined inside constructor or special method) 
    
    - class variable also called statetic variable (class variable defined outside the constructor)

### Methods
    - instance method
    - class method
    - static method

```python
class car():
    def __init__(self):
        self.mil=10 # instance var
        self.com='bmw' # instance var

c1=car()
c2=car()

c2.com='swift' # calling the instace variable
c2.mil=18
print(c1.com,c1.mil)
print(c2.com,c2.mil)
    
```

```python
class car():

    wheels=5 # class variable
    
    def __init__(self):
        self.mil=10 # instance var
        self.com='bmw' # instance var

c1=car()
c2=car()

c2.com='swift' # calling the instace variable
c2.mil=18
c2.wheels=4
print(c1.com,c1.mil,c1.wheels)
print(c2.com,c2.mil,c2.wheels)
    
```

# Inheritance
- it always works with minimum 2 classes
- all properties of parent belongs to child

### Types of Inheritance
1. Single level
2. Multi Level
3. Multiple Inheritance
4. Hybrid Inheritance

```python
class A():
    def feature1(self):
        print('Yunus says Hi! to all')
    def feature2(self):
        print('feature2 says Hi!')
a1=A()
a1.feature1()
a1.feature2()
```

```python
class A():
    def feature1(self):
        print('Yunus says Hi! to all')
    def feature2(self):
        print('feature2 says Hi!')

class B(A): # A class inheriate by class B
    def feature3(self):
        print('Sonalika says Hi!')
    def feature4(self):
        print('Yunus says Good Bye')

        
a1=A()
b1=B()
b1.feature4()
```

```python
class A():
    def feature1(self):
        print('Yunus says Hi! to all')
    def feature2(self):
        print('feature2 says Hi!')

class B(): # A class inheriate by class B
    def feature3(self):
        print('Sonalika says Hi!')
    def feature4(self):
        print('Yunus says Good Bye')

class C(A,B):
    def feature5(self):
        print('Debo belongs to Manish')

        
a1=A()
b1=B()
c1=C()
c1.feature5()
b1.feature3()
c1.feature5()
```
