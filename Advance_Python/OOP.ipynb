# **__init__ methode in python**

```python
# two type of method
    user defined
    special method
```

```python
class computer: #class name
    def __init__(self): # constructor
        print('hello')
        
com1=computer() # in this we don't need to called the method
print(type(com1))
# config is userdefind methode
```

```python
class computer: #class name
    def __init__(self): # constructor
        print('hello')
        
com1=computer()
com2=computer()# in this we don't need to called the method
com1.__init__()
print(type(com1))
# config is userdefind methode
```

```python
class computer(): # class 
    def config(self):  # method
        print('i5',':',8)
    def __init__(self): # special method (constructor)
        print('i7',":",16)
com1=computer()
com1.config()
```
