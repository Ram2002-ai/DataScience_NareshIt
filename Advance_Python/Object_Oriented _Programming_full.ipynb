## METHODS & DIFFERENT TYPES OF METHODS
- OBJECT HAS 2 THINGS -- VARIABLES & METHODS   
- VARIABLES ARE STORE THE DATA 
- METHODS WILL PERFORM THE BEHAVIOUR , TO PERFORM SOME OPERATION THATS WHERE METHOD COMES TO THE PICTURE
- 3 TYPE OF METHODS ( INSTANCE METHOD, CLASS METHOD,  STATIC METHOD)


```python
#lets take student class ,in the student class we have multiple variable like name, roll number & marks. 
#at this point we only focus on marks & below lets create init function cuz i have to declare some variable 
# m1, m2, m3 variable are those are instance variable
# outside of the function is called static variable as we learn now

class student: 
        
    school = 'nit'   #school is class variable
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable (m1,m2,m3 are instance variable)
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
#lets create a method to calculate average of the marks 
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(s2.avg())
```

# Instance Method
- AVERAGE ONLY WE CALLED AS INSTANCE METHOD 
- IN INSTANCE ONLY WE HAVE 2 TYPES

- ACCESSOR & MUTATOR 


- ACCESSOR TYPE - we are only fetching the values (when we fetch the instance variable is called ACCESOR &

- when we modify the variable then we called as MUTATOR 

```python
class Student:
    school='nit' #class/static variable

    def __init__(self,m1,m2,m3):
        self.m1=m1 #instance var
        self.m2=m2
        self.m3=m3

    def avg(ram): # we can put any any word not just self
        return(ram.m1+ram.m2+ram.m3)/3

    def get_m1(self): #getter method - get the value (getter only fetch the value)--Accesor
        return self.m1

    def set_m1(self,value):  #setter method (setter - set the value) - SETTER only set the value -- MUTATOR
        self.m1=value

# lets create object hear as we created s1,s2 we created 2 object &also i want to pass value in this

s1=Student(10,20,30)
s2=Student(5,15,10)

s2.set_m1(89)

print(s1.avg())
print(s2.avg())

print(s2.m1)
```

# Class Method

- instance variable can use in instance method
- school is basically your class variable & lets create class method 
- if you work in instance then you can pass instance method 
- if you work in class then you need to pass class method

```python
class student: 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
#lets create a method to calculate average of the marks 
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    def info(cls): # user need to define cls - for class variable 
        return f"{cls.school}: marks {cls.m1+cls.m2+cls.m3}" 
        
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(s2.avg())



print(s1.info())
print(s2.info())
```
